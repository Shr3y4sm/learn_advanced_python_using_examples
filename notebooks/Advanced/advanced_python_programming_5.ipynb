```python
#import plotly.express as px

# Sample Data
df = px.data.gapminder().query("country == 'India'")

# Create Line Plot
fig = px.line(df, x="year", y="gdpPercap", title="India's GDP Over Time")

# Show Plot
fig.show()
```

```python
# Sample dataset
data = pd.DataFrame({
    'wt': [1.5, 2.3, 3.1, 4.2, 5.0],
    'mpg': [35, 30, 25, 20, 15]
})

# Create interactive scatter plot
fig = px.scatter(data, x='wt', y='mpg', title="Interactive Scatter Plot",
                 labels={'wt': 'Weight (tons)', 'mpg': 'Miles per gallon'})

# Show figure
fig.show()
```

```python
df = pd.read_csv('../../data/mtcars.csv')
df
```

```python
# Create interactive scatter plot
fig = px.scatter(df, x='wt', y='mpg', title="Interactive Scatter Plot",
                 labels={'wt': 'Weight (tons)', 'mpg': 'Miles per gallon'})

# Show figure
fig.show()
```

```python
f = px.scatter(data_frame=df, x='gear', y='mpg')

print(f"Default Width: {f.layout.width}")  # Usually None
print(f"Default Height: {f.layout.height}")  # Usually None

f.update_layout(width=800, height=500)

print(f"Default Width: {f.layout.width}")  # Usually None
print(f"Default Height: {f.layout.height}")  # Usually None

# Show the figure
f.show()# Check default width and height
```

```python
f = px.scatter(data_frame=df, x='gear', y='mpg')

f.show()
```

```python
f = px.bar(data_frame=df, x='mpg', orientation='v')

f.show()
```

```python
f = px.bar(data_frame=df, x='gear', y='mpg', orientation='v')

f.show()
```

```python
f = px.histogram(data_frame=df, x='mpg', facet_row_spacing=0.8)
# Adjust column spacing & Center X-axis labels
f.update_layout(
    width=1000,
    height=600,
    bargap=0.3,  # Adjust spacing between bars
)
f.show()
```

```python
f = px.box(data_frame=df, y=['hp', 'disp'], title='Box Plot using Plotly')
# Adjust column spacing & Center X-axis labels

f.show()
```

```python
f = px.violin(data_frame=df, y=['hp', 'disp'], title='Violin Plot using Plotly', box=True)
# Adjust column spacing & Center X-axis labels

f.show()
```

```python
vc = df.gear.value_counts()
f = px.pie(data_frame=vc, values=vc.values, names=vc.index, title='Pie plot using plotly')
f.show()
```

```python
ax1 = plt.gca()
ax2 = ax1.twinx()

df1 = df.sort_values(by='wt')

ax1.plot(df1.wt, df1.hp, label='hp', color='red')
ax2.plot(df1.wt, df1.disp, label='disp', color='green')


plt.legend()
plt.tight_layout()
plt.show()
```
